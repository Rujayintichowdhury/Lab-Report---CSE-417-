#01.Translation:  



```python
import cv2  # OpenCV library for Python
import numpy as np  # numerical calculations and matrices.
import matplotlib.pyplot as plt

# Upload/read image
image = cv2.imread('/content/images01.jpg')

# Check if image was loaded successfully
if image is None:
    print("Error: Image not found. Please upload 'image.jpg' to /content/ directory.")
else:
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Image dimensions
    h, w = image.shape[:2]  # (height, width, channels)


    # Translation values
    tx = 100
    ty = 50

    # Translation matrix
    M = np.float32([
        [1, 0, tx],
        [0, 1, ty]
    ])

    # Apply transformation
    translated = cv2.warpAffine(image, M, (w, h))

    # Display
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)  # 1 → number of rows, 2 → number of columns, 1 → first position
    plt.imshow(image)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(translated)
    plt.title("Translated Image")
    plt.axis("off")

    plt.show()
```

#02.Scaling of a 2D Image

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('/content/images02.jpg')

# Check if image was loaded successfully
if image is None:
    print("Error: Image not found. Please ensure 'image.jpg' is in the /content/ directory.")
else:
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    h, w = image.shape[:2]

    # Scaling factors
    sx = 1.5
    sy = 1.5

    # Scaling matrix
    M = np.float32(
        [
            [sx, 0, 0],
            [0, sy, 0]
        ]
    )

    scaled = cv2.warpAffine(
        image,
        M,
        (int(w * sx), int(h * sy))
    )

    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(image)
    plt.title("Original Image")
    plt.axis("off")
    plt.subplot(1, 2, 2)
    plt.imshow(scaled)
    plt.title("Scaled Image")
    plt.axis("off")

    plt.show()
```

#03.Rotation of a 2D Image

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('/content/images03.jpg')

# Check if image was loaded successfully
if image is None:
    print("Error: Image not found. Please ensure 'image.jpg' is in the /content/ directory.")
else:
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    h, w = image.shape[:2]

    # Center of image
    center = (w // 2, h // 2)

    # Rotation angle
    angle = 45

    # Rotation matrix
    M = cv2.getRotationMatrix2D(center, angle, 1.0)

    rotated = cv2.warpAffine(
        image,
        M,
        (w, h)
    )

    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(image)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(rotated)
    plt.title("Rotated Image - 45°")
    plt.axis("off")

    plt.show()
```

#06.Combined Transformation

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('/content/images05.jpg')

# Check if image was loaded successfully
if image is None:
    print("Error: Image not found. Please ensure 'image.jpg' is in the /content/ directory.")
else:
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    h, w = image.shape[:2]

    # -------------------------
    # Step 1: Scaling
    # -------------------------

    sx = 1.2
    sy = 1.2

    scaled = cv2.resize(
        image,
        None,
        fx=sx,
        fy=sy
    )

    # -------------------------
    # Step 2: Rotation
    # -------------------------

    h2, w2 = scaled.shape[:2]

    center = (w2 // 2, h2 // 2)

    M_rotation = cv2.getRotationMatrix2D(
        center,
        30,
        1.0
    )

    rotated = cv2.warpAffine(
        scaled,
        M_rotation,
        (w2, h2)
    )

    # -------------------------
    # Step 3: Translation
    # -------------------------

    tx = 50
    ty = 30

    M_translation = np.float32([
        [1, 0, tx],
        [0, 1, ty]
    ])

    transformed = cv2.warpAffine(
        rotated,
        M_translation,
        (w2 + tx, h2 + ty)
    )

    # Display
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(image)
    plt.title("Original")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(rotated)
    plt.title("After Rotation")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(transformed)
    plt.title("Final Transformation")
    plt.axis("off")

    plt.show()
```
