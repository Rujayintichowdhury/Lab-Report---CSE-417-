# Direct Line Drawing Algorithm Using Equation Method

```python
# -*- coding: utf-8 -*-
"""Untitled0.ipynb

Automatically generated by Colab.

Original file is located at
    https://colab.research.google.com/drive/12mnRa19ntz67kQbV9Id6_c0jl_fPazoe
"""

import matplotlib.pyplot as plt

def direct_line(x1, y1, x2, y2):
    pixels = []

    print("\n---- Direct Line Equation Method ----")
    print("Starting point:", (x1, y1))
    print("Ending point:  ", (x2, y2))

    dx = x2 - x1
    dy = y2 - y1

    # Vertical line
    if dx == 0:
        print("Case: Vertical Line\n")
        print("Slope is undefined\n")

        start_y = min(y1, y2)
        end_y = max(y1, y2)

        for y in range(start_y, end_y + 1):
            pixels.append((x1, y))

            print(
                "x =", x1,
                "y =", y,
                "pixels ->", (x1, y)
            )

    # Horizontal or shallow line
    else:
        m = dy / dx
        b = y1 - m * x1

        print("Slope:", m)
        print("Y-intercept:", b)

        if abs(m) <= 1:
            print("\nCase: Shallow Line\n")

            start_x = min(x1, x2)
            end_x = max(x1, x2)

            for x in range(start_x, end_x + 1):
                y = m * x + b
                round_y = round(y)

                pixels.append((x, round_y))

                print(
                    "x =", x,
                    "y =", round_y,
                    "pixels ->", (x, round_y)
                )

        # Steep line
        else:
            print("\nCase: Steep Line\n")

            start_y = min(y1, y2)
            end_y = max(y1, y2)

            for y in range(start_y, end_y + 1):
                x = (y - b) / m
                round_x = round(x)

                pixels.append((round_x, y))

                print(
                    "x =", round_x,
                    "y =", y,
                    "pixels ->", (round_x, y)
                )

    return pixels


# -------------------------------
# User Input
# -------------------------------

x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))

x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))


# Call the function
pixels = direct_line(x1, y1, x2, y2)


# Display final pixels
print("\n---- Final Pixel Coordinates ----")

for pixel in pixels:
    print(pixel)

import matplotlib.pyplot as plt

def plot_pixels(pixels, x1, y1, x2, y2):
    xs = [p[0] for p in pixels]
    ys = [p[1] for p in pixels]

    plt.figure(figsize=(6, 6))


    plt.scatter(xs, ys, s=80)


    plt.plot([x1, x2], [y1, y2], '-', linewidth=1)

    plt.gca().set_aspect('equal', adjustable='box')
    plt.grid(True)
    plt.title("Line Plot (pixels)")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.show()


# after calling direct_line(...)
pixels = direct_line(x1, y1, x2, y2)

print("\n---- Final Pixel Coordinates ----")
for pixel in pixels:
    print(pixel)

# plot the line
plot_pixels(pixels, x1, y1, x2, y2)
```

#DDA (Digital Differential Analyzer) Line Drawing Algorithm in Python using (x1, y1) and (x2, y2):

```python
import matplotlib.pyplot as plt

# Input coordinates
x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))
x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))

# Calculate dx and dy
dx = x2 - x1
dy = y2 - y1

# Calculate steps
steps = max(abs(dx), abs(dy))

# Calculate increments
x_inc = dx / steps
y_inc = dy / steps

# Starting point
x = x1
y = y1

points_x = []
points_y = []

# DDA Algorithm
for i in range(steps + 1):
    points_x.append(round(x))
    points_y.append(round(y))

    x = x + x_inc
    y = y + y_inc

# Display points
print("Line points:")
for i in range(len(points_x)):
    print(points_x[i], points_y[i])

# Draw the line
plt.plot(points_x, points_y, 'ro-')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("DDA Line Drawing Algorithm")
plt.grid(True)
plt.show()
```

#Bresenham's Line and Circle Drawing Algorithm


#1. Bresenham Straight Line

```python
import matplotlib.pyplot as plt

x1, y1 = 2, 3
x2, y2 = 10, 8

dx = abs(x2 - x1)
dy = abs(y2 - y1)

x = x1
y = y1

sx = 1 if x2 > x1 else -1
sy = 1 if y2 > y1 else -1

points_x = []
points_y = []

if dx > dy:
    p = 2 * dy - dx

    for i in range(dx + 1):
        points_x.append(x)
        points_y.append(y)

        if p >= 0:
            y += sy
            p += 2 * (dy - dx)
        else:
            p += 2 * dy

        x += sx

else:
    p = 2 * dx - dy

    for i in range(dy + 1):
        points_x.append(x)
        points_y.append(y)

        if p >= 0:
            x += sx
            p += 2 * (dx - dy)
        else:
            p += 2 * dx

        y += sy

plt.plot(points_x, points_y, 'ro-')
plt.grid()
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Bresenham Line Drawing")
plt.show()
```

2. Bresenham Circle

```python
import matplotlib.pyplot as plt

xc = 0
yc = 0
r = 10

x = 0
y = r
p = 3 - 2 * r

points_x = []
points_y = []

while x <= y:

    points = [
        (xc + x, yc + y),
        (xc - x, yc + y),
        (xc + x, yc - y),
        (xc - x, yc - y),
        (xc + y, yc + x),
        (xc - y, yc + x),
        (xc + y, yc - x),
        (xc - y, yc - x)
    ]

    for px, py in points:
        points_x.append(px)
        points_y.append(py)

    if p < 0:
        p = p + 4 * x + 6
    else:
        p = p + 4 * (x - y) + 10
        y -= 1

    x += 1

plt.scatter(points_x, points_y)
plt.gca().set_aspect('equal')
plt.grid()
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Bresenham Circle Drawing")
plt.show()
```
